#Q1

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
img = mpimg.imread("sample1.jpg")
```

```python
print("Image shape:", img.shape)
```

```python
print("Minimum pixel value:", img.min())
print("Maximum pixel value:", img.max())
```

```python
plt.imshow(img)
plt.axis("off")
plt.title("Original Image")
plt.show()
```

#Q2

```python
R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(R, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(G, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(B, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.show()
```

```python
recombined = np.stack([R, G, B], axis=-1)
plt.imshow(recombined)
plt.axis("off")
plt.title("Recombined Image")
plt.show()
```

#Q3

```python
gray_manual = 0.299 * R + 0.587 * G + 0.114 * B
plt.imshow(gray_manual, cmap="gray")
plt.title("Manual Grayscale")
plt.axis("off")
plt.show()
```

```python
gray_avg = np.mean(img, axis=2)
plt.imshow(gray_avg, cmap="gray")
plt.title("Average Grayscale")
plt.axis("off")
plt.show()
```

```python
def blur_gray(img):
  kernel = np.ones((3,3))/9
  H,W = img.shape
  padded = np.pad(img, pad_width=1, mode='constant', constant_values=0)
  blurred = np.zeros((H,W))
  for i in range(H):
    for j in range(W):
      region = padded[i:i+3, j:j+3]
      blurred[i,j] = np.sum(region*kernel)
  return blurred
blurred_img = blur_gray(gray_manual)
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(gray_manual, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(blurred_img, cmap="gray")
plt.title("Blurred Image")
plt.axis("off")

plt.show()
```

1.   A convolution kernel defines how each pixel in an image is combined with its neighboring pixels to produce a new pixel value.
2.   Changing the kernel changes the type of filtering applied to the image, such as blurring, sharpening, or detecting edges.
3. Grayscale conversion uses weighted averages because the human eye is more sensitive to green light and less sensitive to blue light, so weighted values better represent perceived brightness.
